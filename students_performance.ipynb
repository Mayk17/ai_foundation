```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("StudentsPerformance.csv")
print("Dataset shape(row, columns):", df.shape) # qancha qator va qancha ustun borligin bilish uchun kerak
print("Column name:", df.columns.tolist()) # ustundagi labalellar nomi

print("Frist 5 rows of the dataset:")
print(df.head()) # 5ta ustunni olib beradi

print("Summary statistics:")
print(df.describe()) # stastitik malumot

print("data types and null info:")
print(df.info()) #
```

```python
print("Average score by subject:")
df[["Age", "Absences", "StudyTimeWeekly"]].mean()
```

```python
print("Maximum score by students:")
df[["Age", "Absences", "StudyTimeWeekly"]].max()
```

```python
print("Minimu score by students:")
df[["Age", "Absences", "StudyTimeWeekly"]].min()
```

```python
print("Correlation")
df[["Age", "Absences", "StudyTimeWeekly"]].corr()
```

```python
test_prep_groupby = df.groupby('test preparation course')[["math score", "reading score", "writing score"]].mean()
print("student score by test preparation course:")
print(test_prep_groupby)

# Gender-based  average
gender_avg = df.groupby('gender')[["math score", "reading score", "writing score"]].mean()
print("student score by gender:")
print(gender_avg)

#who passed vs failed

pass_rate = df["passed"].value_counts(normalize=True) * 100
print("Pass rate:\n", pass_rate)
```

```python
print(df.columns)
```

```python
#Convert scores to Numpy array
scores = df[["math score", "reading score", "writing score"]].values


# Mean, Std Devition
print("Numpy Mean:", np.mean(scores, axis=0))
print("Numpy Std Deviation:", np.std(scores, axis=0))
#Z-score (normalize)
z_scores = (scores - np.mean(scores, axis=0)) / np.std(scores, axis=0)
print("Z-scores:\n", z_scores)

# Average 0
# Negative -
# Positive +
```

```python
# How many students scored above 90 average?
top_scorers = df[df['math score'] >= 90]
print("Top Scorers above 90:", len(top_scorers))

#How many failed ?
failed_students = df[df['math score'] <=60]
print("Failed Students:", len(failed_students))
```
