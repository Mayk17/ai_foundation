```python
import numpy as np

arr = np.array([1, 3, 5, 6])
print('numpy_array:', arr)
arr.shape
```

```python
arr2 = np.array([[1, 4, 5, 6], [3, 5, 7, 5]])
print("numpy_array:", arr2)
arr2.shape
```

```python
#dtype
a = np.array([1, 2, 4])
print(a.dtype)
```

```python
#reshape()
import numpy as np
a= np.array([1, 2, 5, 6, 9, 3])
b = a.reshape(2, 3)
print("shaped_array:", b)
```

```python
#astype
a = np.array([1, 2, 4,])
b = a.astype(float)
print("float:", b)
```

```python
# np.zeos(), np.ones() => dummy data # hammasi matrisa hosil qilib berdi.
z = np.zeros((2, 4))
o = np.ones((2, 4))

print("zeros:", z)
print("ones:", o)
```

```python
#ndim
a = np.array([[1, 2], [4, 6]])
print("ndim:", a.ndim)
```

```python
#arange()
a = np.arange(1, 7)
print("arange:", a)
```

```python
# sum() mean() mix() max()
import numpy as np
a = np.array([1, 2, 3, 4])

print("sum:", a.sum())
print("mean:", a.mean())
print("max:", a.max())
print("min:", a.min())
```

```python
# flatten()

a = np.array([[11, 2], [4, 6]])
print("flatten:", a.flatten())
```

```python
# T (Transpose)
a = np.array([[1, 4], [5, 7]])
print("transpose:", a.T)
```

```python
#dot() @
a = np.array([[1, 2], [7, 0]])
b = np.array([[2, 3], [5, 1]])
print("dot/ multiplication:", a.dot(b))
```

```python
#random.rand()
a = np.random.rand(2, 4) # => 0.0 and 1.0 / 0.0 included/ 1.0 excluded
print("random rand:", a)
```
